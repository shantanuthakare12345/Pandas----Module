```python
#Create an object of DataFrame By using List in List/ tuple in tuple
lst=[[10,20,30,40],["RS","TR","DR","JH"]]
df=pd.DataFrame(lst)
print("-"*50)
print("Content of DataFrame")
print(df,type(df))
print("-"*50)
```

```python
#Create an object of DataFrame By using List in List/ tuple in tuple
lst=[[10,20,30,40],["RS","TR","DR","JH"]]
df=pd.DataFrame(lst,index=["ID1","ID2"],columns=["Rec"+str(i) for i in range(1,len(lst[0])+1)])
print("-"*50)
print("Content of DataFrame")
print(df,type(df))
print("-"*50)
```

```python
#Create DataFrame by using ndarray Object
a=np.array([[10,20,30],[40,50,60],[70,80,90]])
df=pd.DataFrame(a)
print("-"*50)
print("Content of DataFrame")
print("-"*50)
print(df,type(df))
print("-"*50)
```

```python
#Create DataFrame by using ndarray Object
a=np.array([[10,20,30],[40,50,60],[70,80,90]]).flatten()
df=pd.DataFrame(a)
print("-"*50)
print("Content of DataFrame")
print("-"*50)
print(df,type(df))
print("-"*50)
```

```python
#Create DataFrame by using ndarray Object
a=np.array([[10,20,30],[40,50,60],[70,80,90]]).flatten()
s=pd.Series(a)
df=pd.DataFrame(s)
print("-"*50)
print("Content of DataFrame")
print("-"*50)
print(df,type(df))
print("-"*50)
```
