```python
#Create an Object Series with Programmer Defined Indices
import pandas as pd
import numpy as np
s=pd.Series(["Rossum","Travis","Kinney","Dennis","Ritche"],
 index=["Name"+str(i) for i in range(1,len(["Rossum","Travis","Kinney","Dennis","Ritche"])+1)])
print("Content of Series Object")
print(s,type(s))
```

```python
#Get the Value from Series Object By using Programmer-Defined Index
#Get Name3 from series Object
print(s["Name3"])
```

```python
#Get Name5 from series Object
print(s["Name5"])
```

```python
#Get the Value from Series Object By using Programmer-Defined Random Indices
#Get Name1,Name3 and Name5 from series Object
print(s[["Name1","Name3","Name5"]])
```

```python
#Get the Values from Series Object By using Programmer-Defined Slicing
#Get Name1 to Name4 from series Object
print(s["Name1":"Name4"])
```

```python
print(s["Name1":"Name5":2])
```

```python
#Setting the Value(s) to the Series Object
#Create an Object Series with Default Indices
import pandas as pd
import numpy as np
s=pd.Series(["Rossum","Travis","Kinney","Dennis","Ritche"])
print("Content of Series Object")
print(s,type(s))
```

```python
#1. Set OR Modify the Value of Series Object By using Indexing
#Set OR Update 0th Index Name as "Guido Van Rossum"
s[0]="Guido Van Rossum"
print(s,type(s),id(s))
```

```python
#Set OR Update 2nd Index Name as "WES MC KINNEY"
s[2]="WES MC KINNEY"
print(s,type(s),id(s))
```

```python
#2. Set OR Modify the Values of Series Object By using Random Indexing
#Update 1st and 4th Index names
s[[1,4]]=["Travis Oliphant","Dennis Ritche"]
print(s,type(s),id(s))
```

```python
#3. Set OR Modify the Values of Series Object By using Slicing
#Update 0th to 2th Index names
s[0:3]=["Rossum","Travis","Kinney"]
print(s,type(s),id(s))
```
