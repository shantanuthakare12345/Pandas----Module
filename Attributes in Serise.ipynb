```python
#create an object Serice with Default Indices
import pandas as pd
import numpy as np
s=pd.Series(["Rossum","Travis","Kinney","Dennis","Ritche"],
           index=['Name'+str(i) for i in range(1,len(["Rossum","Travis","Kinney","Dennis","Ritche"])+1)])
print("Content of Series Object")
print(s,type(s))
```

```python
#Find the Dimension of Series Object by using 'ndim'
print("Dimension of Series Object s=",s.ndim)
```

```python
#Find the Data type of Series Object by using 'dtype'
print("Data Type of Series Object=",s.dtype)
```

```python
##Find the size and shape of Series Object by using 'size' and 'shape'
print("Number of Elements in Series Object =",s.size)
print("Shape of series object =",s.shape)
```
